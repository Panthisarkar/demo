```python
import librosa
import soundfile as sf
import os

# Path to the folder containing the audio files
folder_path = '/content/drive/MyDrive/aawaj/data/english/test/'  # Update with the correct path

# Desired sampling rate
new_sample_rate = 16000

# Loop through all files in the folder
for filename in os.listdir(folder_path):
    if filename.endswith('.wav'):  # Process only .wav files
        audio_path = os.path.join(folder_path, filename)

        # Load the audio file and resample to the desired rate
        y, sr = librosa.load(audio_path, sr=new_sample_rate)

        # Save the resampled audio, overwriting the original file
        sf.write(audio_path, y, sr)

        print(f"Resampled and replaced: {filename}")
```
